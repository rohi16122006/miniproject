# 01: Explore VitalDB
Inspect `clinical_data.csv`, then plot one downloaded case. Run `python scripts/download_vitaldb.py` first.

```
import pandas as pd, numpy as np, matplotlib.pyplot as plt
df = pd.read_csv('../data/raw/vitaldb/clinical_data.csv')
print(df.shape)
df[['caseid','age','sex','aline1','intraop_phe','caseend']].head()
```

```
# Cases with an arterial line and a phenylephrine dose
ok = df[df.aline1.notna() & (df.intraop_phe > 0)]
print(len(ok), 'cases')
```

```
# Plot one saved case (columns: sbp, dbp, hr, drug_rate; 10 s per sample)
d = np.load('../data/raw/vitaldb/case_0984.npz')['values']
t = np.arange(len(d)) * 10 / 60
fig, ax = plt.subplots(2, 1, sharex=True, figsize=(10, 5))
ax[0].plot(t, d[:, 0], label='SBP'); ax[0].plot(t, d[:, 1], label='DBP'); ax[0].legend(); ax[0].set_ylabel('mmHg')
ax[1].plot(t, d[:, 3], color='tab:red'); ax[1].set_ylabel('phenylephrine rate (mL/hr)'); ax[1].set_xlabel('minutes')
plt.show()
```
