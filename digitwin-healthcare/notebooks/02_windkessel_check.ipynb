# 02: Windkessel sanity and sensitivity check
How SBP/DBP respond to resistance R and compliance C (Step 4 sensitivity test).

```python
import sys; sys.path.insert(0, '../backend')
import numpy as np, matplotlib.pyplot as plt
from app.physics.windkessel import simulate_beat
sv, hr = 66.0, 75
for R in (0.8, 1.1, 1.4):
    Cs = np.linspace(0.5, 3, 12)
    sbp = [simulate_beat(R, C, hr, sv).sbp for C in Cs]
    dbp = [simulate_beat(R, C, hr, sv).dbp for C in Cs]
    plt.plot(Cs, sbp, label=f'SBP R={R}'); plt.plot(Cs, dbp, '--', label=f'DBP R={R}')
plt.xlabel('compliance C (ml/mmHg)'); plt.ylabel('mmHg'); plt.legend(); plt.show()
```
